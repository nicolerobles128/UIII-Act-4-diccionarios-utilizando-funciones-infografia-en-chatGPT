#### Diccionario Militar

```python
# Diccionario vacío
diccionario_militar = {}

# Contador para IDs
contador = 1

# -------- FUNCIONES --------

def agregar_militar():
    global contador

    print("\n--- INGRESAR DATOS DEL MILITAR ---")

    rango = input("Rango: ")
    batallon = input("Batallón: ")
    especialidad = input("Especialidad de combate: ")
    misiones = input("Misiones cumplidas: ")
    condecoraciones = input("Condecoraciones: ")
    base = input("Base asignada: ")

    diccionario_militar[contador] = {
        "Rango": rango,
        "Batallon": batallon,
        "Especialidad": especialidad,
        "Misiones": misiones,
        "Condecoraciones": condecoraciones,
        "Base": base
    }

    contador += 1
    print("\nMilitar agregado correctamente.\n")


def listar_militares():
    print("\n----- LISTA DE MILITARES -----")

    print("{:<5} {:<12} {:<12} {:<20} {:<10} {:<18} {:<12}".format(
        "ID", "Rango", "Batallon", "Especialidad", "Misiones", "Condecoraciones", "Base"
    ))

    print("-" * 95)

    for key, value in diccionario_militar.items():
        print("{:<5} {:<12} {:<12} {:<20} {:<10} {:<18} {:<12}".format(
            key,
            value["Rango"],
            value["Batallon"],
            value["Especialidad"],
            value["Misiones"],
            value["Condecoraciones"],
            value["Base"]
        ))
    print()


def buscar_militar():
    id_buscar = int(input("Ingrese ID del militar a buscar: "))

    if id_buscar in diccionario_militar:
        print("\nMilitar encontrado:\n")

        print("{:<5} {:<12} {:<12} {:<20} {:<10} {:<18} {:<12}".format(
            "ID", "Rango", "Batallon", "Especialidad", "Misiones", "Condecoraciones", "Base"
        ))

        print("-" * 95)

        for key, value in diccionario_militar.items():
            if key == id_buscar:
                print("{:<5} {:<12} {:<12} {:<20} {:<10} {:<18} {:<12}".format(
                    key,
                    value["Rango"],
                    value["Batallon"],
                    value["Especialidad"],
                    value["Misiones"],
                    value["Condecoraciones"],
                    value["Base"]
                ))
    else:
        print("No se encontró el militar.")


def borrar_militar():
    id_borrar = int(input("Ingrese ID del militar a borrar: "))

    if id_borrar in diccionario_militar:
        del diccionario_militar[id_borrar]
        print("Militar eliminado correctamente.")
    else:
        print("No se encontró el militar.")


def actualizar_militar():
    id_actualizar = int(input("Ingrese ID del militar a actualizar: "))

    if id_actualizar in diccionario_militar:
        print("\n--- NUEVOS DATOS ---")

        rango = input("Nuevo rango: ")
        batallon = input("Nuevo batallón: ")
        especialidad = input("Nueva especialidad: ")
        misiones = input("Nuevas misiones: ")
        condecoraciones = input("Nuevas condecoraciones: ")
        base = input("Nueva base: ")

        diccionario_militar[id_actualizar] = {
            "Rango": rango,
            "Batallon": batallon,
            "Especialidad": especialidad,
            "Misiones": misiones,
            "Condecoraciones": condecoraciones,
            "Base": base
        }

        print("Militar actualizado correctamente.")
    else:
        print("No se encontró el militar.")


# -------- MENU --------

while True:
    print("\n------ MENU ------")
    print("1. Agregar militar")
    print("2. Listar militares")
    print("3. Buscar militar")
    print("4. Borrar militar")
    print("5. Actualizar militar")
    print("6. Salir")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        agregar_militar()
    elif opcion == "2":
        listar_militares()
    elif opcion == "3":
        buscar_militar()
    elif opcion == "4":
        borrar_militar()
    elif opcion == "5":
        actualizar_militar()
    elif opcion == "6":
        print("Saliendo del programa...")
        break
    else:
        print("Opción inválida")
        print("Programa realizado por Nicole Robles = 46")
```
